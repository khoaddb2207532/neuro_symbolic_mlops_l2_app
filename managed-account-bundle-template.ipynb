# Bundle output của một tài khoản Kaggle
Add Input các notebook run thuộc tài khoản này trước khi chạy.

```python
ACCOUNT_ID = "account-1"
EXPECTED_RUN_IDS = [
    "culture-a__mobilenetv3_small__db__seed_42",
    "culture-a__mobilenetv3_small__db__seed_44",
    "culture-a__mobilenetv3_small__db__seed_46",
]
GIT_COMMIT = "REPLACE_GIT_COMMIT"
```

```python
import subprocess
from pathlib import Path
from kaggle_secrets import UserSecretsClient
repo = Path('/kaggle/working/neuro_symbolic_mlops_l2_app')
if not repo.exists():
    token = UserSecretsClient().get_secret('GITHUB_TOKEN')
    subprocess.run(['git', 'clone', '-b', 'rollback', '--single-branch', f'https://{token}@github.com/khoaddb2207532/neuro_symbolic_mlops_l2_app.git', str(repo)], check=True)
subprocess.run(['git', '-C', str(repo), 'fetch', '--all'], check=True)
subprocess.run(['git', '-C', str(repo), 'checkout', GIT_COMMIT], check=True)
```

```python
cmd = [
    'python', '-m', 'pipelines.bundle_account_outputs',
    '--input-root', '/kaggle/input',
    '--output-dir', f'/kaggle/working/bundle_{ACCOUNT_ID}',
    '--account-id', ACCOUNT_ID,
    '--expected-run-ids', *EXPECTED_RUN_IDS,
]
subprocess.run(cmd, cwd=repo, check=True)
```
